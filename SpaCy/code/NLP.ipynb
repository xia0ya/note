# SpaCy自然语言处理

## 初识

### nlp对象

```python
import spacy
# 创建一个空的中文处理流程
nlp = spacy.blank("zh")
# 使用nlp对象处理文本
text = "这是一个例子"
doc = nlp(text)

# 输出举例结果
for token in doc:
    print(token.text)

```

```python
import spacy
# python -m spacy download zh_core_web_sm
nlp = spacy.load("zh_core_web_sm")
doc = nlp("这是一个例子")

for token in doc:
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
# 文本，词元，词性，依存关系，实体类型
```
